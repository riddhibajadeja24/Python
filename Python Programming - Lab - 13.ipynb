### 05) WAP to generate 3 random integers between 100 and 999 which is divisible by 5.

```python
import random

count = 0
while count < 3:
    num = random.randint(100, 999)
    if num % 5 == 0:
        print(num)
        count += 1
```

### 06) WAP to generate 100 random lottery tickets and pick two lucky tickets from it and announce them as Winner and Runner up respectively.

```python
import random

tickets = random.sample(range(1000, 9999), 100)

winner = random.choice(tickets)
runner = random.choice(tickets)

print("Winner:", winner)
print("Runner-up:", runner)
```

### 08) Subtract a week (7 days) from a given date in Python.

```python
import datetime

date = datetime.datetime(2025, 3, 10)
new_date = date - datetime.timedelta(days=7)

print("New Date:", new_date)
```

### 09) WAP to Calculate number of days between two given dates.

```python
import datetime

d1 = datetime.date(2025, 3, 1)
d2 = datetime.date(2025, 3, 10)

diff = d2 - d1
print("Days:", diff.days)
```
