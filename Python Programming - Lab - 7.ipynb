### 03) WAP to find tuples which have all elements divisible by K from a list of tuples.
**Sample Input:**
```
tuple_list = [(2, 4, 6), (3, 6, 9), (4, 8, 12), (5, 10, 15)]
K = 2

```
**Sample Output:**
```
Tuples divisible by 2 : [(2, 4, 6), (4, 8, 12)]
```

````python
tuple_list=[(2,4,6),(3,6,9),(4,8,12),(5,10,15)]
tp=[]
n=int(input("enter divisible "))
for i in tuple_list:
    flag=True
    for j in i:
        if j%n!=0:
            flag=False
    if flag:
        tp.append(i)
print(f"Tuples divisible by {n} :",tp)        
````

### 14) WAP to create (index, value) pairs for prime numbers.

**Sample Input:**
```
T = (4, 7, 9, 11, 15)
```
**Sample Output:**
```
((1, 7), (3, 11))
```

````python
t=(4, 7, 9, 11, 15)
li=[]
for i in range(0,len(t)):
    flag=1
    num=t[i]
    for j in range(2,num):
        if num%j==0:
            flag=0
            break
    if flag==1:
        li.append((i,t[i]),)
            
tuple(li)
    
````

### 15) WAP to split a tuple into even-index and odd-index tuples.

```
T = ('a', 'b', 'c', 'd', 'e')
```

```
even = ('a', 'c', 'e')
odd = ('b', 'd')
```

### 16) WAP to compute column-wise sum from a tuple of lists.

**Sample Input:**
```
T = ([1, 2, 3], [4, 5, 6], [7, 8, 9])
```

**Sample Output:**
```
12 15 18

```

````python
t=([1, 2, 3], [4, 5, 6], [7, 8, 9])
s1=0
s2=0
s3=0
for a,b,c in t:
    s1+=a
    s2+=b
    s3+=c
print(s1,s2,s3)
````
