### 10) WAP to write first 100 prime numbers to a file named primenumbers.txt 
(Note: each number should be in new line)

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n/2) + 1):
        if n % i == 0:
            return False
    return True

count = 0
num = 2

file = open("primenumbers.txt", "w")

while count < 100:
    if is_prime(num):
        file.write(str(num) + "\n")  
        count += 1
    num += 1

file.close()

print("First 100 prime numbers written to primenumbers.txt")
```

### 12) WAP to replace word1 by word2 of a text file. Write the updated data to new file.

```python
word1 = input("Enter word to replace: ")
word2 = input("Enter new word: ")
f1 = open("file1.txt", "r")
data = f1.read()

# Replac
data = data.replace(word1, word2)

# Open new file
f2 = open("output.txt", "w")

# Write updated content
f2.write(data)

# Close files
f1.close()
f2.close()

print("Replacement done successfully")
```
