### 14) WAP to check whether the given number is Palindrome or not.

```python
n=int(input("enter number "))
rev=0
temp=n
while temp>0:
    rem=temp%10
    rev=rev*10+rem
    temp=temp//10
if n==rev:
    print("palindrome")
else:
    print("not palindrome")
```
