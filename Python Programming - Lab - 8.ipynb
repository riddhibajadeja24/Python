### 11) WAP to find the student with the highest total marks, given student names and their marks.

**Sample Input:**
```
{'Amit': [70, 80, 90], 'Riya': [85, 75, 80], 'Neha': [60, 70, 65]}
```
**Sample Output:**
```
Topper: Amit
```

````python
students = {
    'Amit': [70, 80, 90],
    'Riya': [85, 75, 80],
    'Neha': [60, 70, 65]
}

topper = ""
max_total = 0

for name, marks in students.items():
    total = sum(marks)
    
    if total > max_total:
        max_total = total
        topper = name

print("Topper:", topper)
````
