### 1) Character Category Counter
**Problem:** Count uppercase, lowercase, digits, and special characters in a string.

**Sample Inputs & Outputs:**
- Input: `Python3@Lab` → Uppercase: 2, Lowercase: 7, Digits: 1, Special: 1
- Input: `HELLO123` → Uppercase: 5, Lowercase: 0, Digits: 3, Special: 0
- Input: `hi@#9` → Uppercase: 0, Lowercase: 2, Digits: 1, Special: 2

```python
s=input("enter string")
a=0
b=0
c=0
d=0
for i in s:
     if i.isupper():
         a+=1
     elif i.islower():
         b+=1
     elif i.isdigit():
         c+=1
     else:
         d+=1
print(f"Uppercase {a},Lowercase {b},Digit {c},Special {d}")
```

### 9) String Compression
- `aaabbcccc` → `a3b2c4`
- `xxxyyyz` → `x3y3z1`
- `abcd` → `a1b1c1d1`

```python
s=input("enter string ")
count=1
a=""
for i in range(1,len(s)):
   if s[i]==s[i-1]:
      count+=1
   else:
       a+=s[i-1]+str(count)
       count=1
a+=s[i-1]+str(count)
a
```

### 10) Toggle Case (without using str.swapcase())
- `PyThOn` → `pYtHoN`
- `HELlo` → `helLO`
- `Python` → `pYTHON`

```python
s=input("enter string ")
li=""
for i in s:
    if i.isupper():
        li+=i.lower()
    elif i.islower():
        li+=i.upper()
    else:
        li+=i
print(li)
```
