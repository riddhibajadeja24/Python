### 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
def isprime(n):
    if n<=1:
        return 0
    for i in range(2,n):
        if n%i==0:
            return 0
    return 1
isprime(6)
```

### 04) Write a function that returns the list of Prime numbers between given two numbers.

```python
def prime_num(a,b):
    li=[]
    for i in range(a,b+1):
         for j in range(2,i):
             if i%j==0:
                 break
         else:
             li.append(i)
    return li
print(prime_num(10,20))
```

### 05) Write a function that returns True if the given string is Palindrome or False otherwise.

```python
def palindrome(n):
    if n==n[::-1]:
        return True
    else:
        return False
palindrome("jadeja")
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(5))
```

### 10) Write a function to get the sum of the scores ending with zero.
#### Example : scores = [200, 456, 300, 100, 234, 678] 
#### Ans = 200 + 300 + 100 = 600

```python
li = [200, 456, 300, 100, 234, 678]
def scores_zero(li):
    sum=0
    for i  in li:
        if i%2==0 and i%5 ==0:
            sum+=i
    return sum
scores_zero(li)
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
dic={'a': 10, 'b':20, 'c':30, 'd':40}
def fn_dict(dic):
    dic1={}
    for i,j in dic.items():
        dic1[j]=i
    return dic1
fn_dict(dic)
```

### 12) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python

def latters(s1):
    upper=0
    lower=0
    for i in s1:
        if i.isupper():
            upper+=1
        else:
            lower+=1
    return print(upper,lower)
s1='AbcDEfgh'
latters(s1)
```

### 16) Write udfs to call the functions with following types of arguments:
1. Positional Arguments
2. Keyword Arguments
3. Default Arguments
4. Variable Legngth Positional(*args) & variable length Keyword Arguments (**kwargs)
5. Keyword-Only & Positional Only Arguments

```python
# positional Arguments
def add(a,b):
    return a+b
add(3,6)
```

```python
# keyword arguments
def detail(name,age):
    return print(name,age)
detail(age=20,name='vishu')
```

```python
# default arguments
def greet(name='ridhu'):
    return print(name)
greet()
```

```python
# variable length positional argments
def total(*args):
    sum=0
    for i in args:
        sum+=i
    return sum
total(7,12,2005)
```

```python
# variable length keyword arguments
def details(**kwargs):
    for i,j in kwargs.items():
        print(i,':',j)
details(name='vishu',age='21',city='jamnagar')
```

```python
# positinal arguments
def divide(a,b,/):
    return a/b
divide(10,2)
```

```python
# keyword only arguments
def login(*,pasw,uname):
    print(pasw,uname)
```
