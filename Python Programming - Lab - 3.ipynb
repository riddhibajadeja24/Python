### 08) WAP to read marks of five subjects. Calculate percentage and print class accordingly. 
#### Fail below 35 
#### Pass Class between 35 to 45 
#### Second Class
#### between 45 to 60
#### First Class between 60 to 70
#### Distinction if more than 70

```python
n1=int(input("enter marks 1:"))
n2=int(input("enter marks 2:"))
n3=int(input("enter marks 3:"))
n4=int(input("enter marks 4:"))
n5=int(input("enter marks 5:"))
sum=n1+n2+n3+n4+n5
print(sum)
avg=float(sum/5)
print(avg)
if (avg>70):
    print("Distinction")
elif (avg<=70 and avg>60):
    print("first class")
elif (avg<=60 and avg>45):
    print("second class")
elif(avg<=45 and avg>35):
    print("pass")
else:
    print("fail")
```
